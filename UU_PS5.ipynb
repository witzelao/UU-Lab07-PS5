```python
ames['Central.Air'] = ames['Central.Air'].map({'Y': 1, 'N': 0})
ames['Central.Air'].mean()
#The estimated proportion of homes with central air is 0.933.
```

```python
rng = np.random.default_rng(42)

R = 2000
results = []

for t in range(R):
    idx = rng.integers(0, len(ames), size=len(ames))
    prop = ames['Central.Air'].iloc[idx].mean()
    results.append(prop)

#The rng seed I used was 42. I used 2000 bootstrap replications.

mean = np.mean(results)
standard_error = np.std(results, ddof=1)
standard_error

#The standard error is 0.00459.
```

```python
print(np.quantile(results, [0.05, 0.95]))
print(np.quantile(results, [0.025, 0.975]))
print(np.quantile(results, [0.005, 0.995]))
```

Ho: p = 0.9
Ha: p =/= 0.9
At every significance level a=0.1, a=0.05, and a=0.01, we reject the null hypothesis that p = 0.9. Every confidence interval excludes p=0.9.

```python
area = np.arcsinh(ames['area'])
price = np.arcsinh(ames['price'])
cor = np.corrcoef(area, price)[0,1]

corrs = []
for t in range(R):
    idx = rng.integers(0, len(ames), size=len(ames))

    area_boot = area.iloc[idx]
    price_boot = price.iloc[idx]
    corr_boot = np.corrcoef(area_boot, price_boot)[0,1]
    corrs.append(corr_boot)

#The rng seed I used was 42. I used 2000 bootstrap replications.

stder = np.std(corrs, ddof=1)
stder

#The standard error is 0.01035
```

```python
print(np.quantile(corrs, [0.05, 0.95]))
print(np.quantile(corrs, [0.025, 0.975]))
print(np.quantile(corrs, [0.005, 0.995]))
```

Ho: p = 0. Ha: p =/= 0. 
At every significance level a=0.1, a=0.05, and a=0.01, we reject the null hypothesis that there is no correlation between area and price. 

Area and price must remain paired during resampling because of their correlation. If the bootstrap randomly assigns area and price without their pairs, this would alter the connection between each house and its price and create unstability in the data. 

```python
meta['Overall Survival (Months)'].mean()
```

```python
rng = np.random.default_rng(42)

R = 2000
results = []

for t in range(R):
    idx = rng.integers(0, len(meta), size=len(meta))
    mean = meta['Overall Survival (Months)'].iloc[idx].mean()
    results.append(mean)

#The rng seed I used was 42. I used 2000 bootstrap replications.

standard_error = np.std(results, ddof=1)
print(standard_error)

#The standard error is 2.1358.

print(np.quantile(results, [0.05, 0.95]))
print(np.quantile(results, [0.025, 0.975]))
print(np.quantile(results, [0.005, 0.995]))
```

Ho: u = 120. Ha: u =/= 120. 
At every significance level a=0.1, a=0.05, and a=0.01, we reject the null hypothesis that the mean survival in months is 120. 

The estimate of 129 months means that the average tested patient lived for 129 months. The confidence intervals represent the interval in which the true average months of survival lies, each one specifically increasing as our confidence increases. 

These results do not perfectly estimate unrestricted mean lifetime, as patients still alive have not been properly recorded. These results estimate the survival of the deceased patients.

```python
subset = data[
    (data['AgeInYearsAtScreening'] >= 18) &
    (data['SleepHoursWeekdaysOrWorkdays'].notna())
]

nhanes = subset['SleepHoursWeekdaysOrWorkdays']

nhanes.mean()
```

```python
rng = np.random.default_rng(42)

R = 2000
results = []

for t in range(R):
    idx = rng.integers(0, len(nhanes), size=len(nhanes))
    mean = nhanes.iloc[idx].mean()
    results.append(mean)

#The rng seed I used was 42. I used 2000 bootstrap replications.

standard_error = np.std(results, ddof=1)
print(standard_error)

#The standard error is 0.02275.

print(np.quantile(results, [0.05, 0.95]))
print(np.quantile(results, [0.025, 0.975]))
print(np.quantile(results, [0.005, 0.995]))
```

Ho: u = 8
Ha: u =/= 8
At every significance level a=0.1, a=0.05, and a=0.01, we reject the null hypothesis that the average hours of sleep in adults in 8. 

Seeing how massive the sample size is, any difference is statistically significant. Practically, these differences are not significant because of how little this actually affects people. 

The results of this study aren't actually applicable to the United States or any general population because of how much the rules of proper sampling are ignored. This data can only be applied to the subjects in the sample data.
